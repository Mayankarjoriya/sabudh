```python
import numpy as np
```

```python
def gradient_descent(X, y, k, tau, learning_rate):

    # Number of training examples
    n = X.shape[0]

    # Number of coefficients
    num_features = X.shape[1]

    # Initialize beta with random values
    beta = np.random.randn(num_features, 1)

    # Previous cost
    previous_cost = float("inf")

    # One loop for gradient descent
    for _ in range(k):

        # Step 1: Calculate predictions
        y_pred = X @ beta

        # Step 2: Calculate error
        error = y_pred - y

        # Step 3: Calculate cost
        cost = np.sum(error ** 2) / (2 * n)

        # Step 4: Check convergence
        if abs(previous_cost - cost) < tau:
            break

        # Step 5: Calculate gradient
        gradient = (X.T @ error) / n

        # Step 6: Update beta
        beta = beta - learning_rate * gradient

        # Step 7: Save current cost
        previous_cost = cost

    return beta, cost
```
